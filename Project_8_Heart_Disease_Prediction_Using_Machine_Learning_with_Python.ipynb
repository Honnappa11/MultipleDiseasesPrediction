**Importing the Dependencies**

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn import svm
```

**Data Collection and Processing**

```python
# loading the csv data to a Pandas DataFrame
data=pd.read_csv('/content/heart.csv')
```

```python
# print first 5 rows of the Dataset
data.head()
```

```python
# print last 5 rows of the dataset
data.tail()
```

```python
# number of rows and columns in the dataset
data.shape
```

```python
# getting some information about the data
data.info()
```

```python
# checking for missing values
data.isnull().sum()
```

```python
# statistical measure about the data
data.describe()
```

```python
# checking the distribution of Target Variable
data['output'].value_counts()
```

1 --> Defective Heart

0 --> Healty Heart

**Splitting the Features and Target**

```python
x=data.drop(columns='output',axis=1)
y=data['output']
```

```python
print(x)
print(y)
```

**Splitting the Data into Training data and Testing data**

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=2)
```

```python
print(x.shape,x_train.shape,x_test.shape)
```

**Model Training**

**SVM**

```python
model=svm.SVC(kernel='linear')
```

```python
# training the svm model with Training data
model.fit(x_train,y_train)
```

**Model Evaluation**

**Accuracy Score**

```python
# accuracy on training data
x_train_prediction=model.predict(x_train)
x_train_accuracy=accuracy_score(x_train_prediction,y_train)
```

```python
print('Accuracy on Training data : ',x_train_accuracy)
```

```python
# accuracy on testing data
x_test_prediction=model.predict(x_test)
x_test_accuracy=accuracy_score(x_test_prediction,y_test)

```

```python
print('Accuracy on Testing data : ',x_test_accuracy)
```

**Building a Predictive System**

```python
input_data=(44,1,0,120,169,0,1,144,1,2.8,0,0,1)

# change the input data to a numpy array
input_data_asarray=np.asarray(input_data)

# reshape the numpy array as we are prediction
input_data_reshaped=input_data_asarray.reshape(1,-1)

prediction=model.predict(input_data_reshaped)
print(prediction)


if prediction[0]==0:
  print('The Person does not have a Heart Disease')
else:
  print('The Person has Heart Disease')
```
